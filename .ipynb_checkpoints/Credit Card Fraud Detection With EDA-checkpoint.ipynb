```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('always') 
```

```python
pd.set_option('display.max_columns', None)
df = pd.read_csv("fraudTrain.csv")
df.head(10)
```

## Data Cleaning

```python
df.rename(columns={'dob': 'age', 'amt': 'amount'}, inplace=True)
```

```python
df['first'] = df['first'] + ' ' + df['last']
df.drop(['last'], axis=1, inplace=True)
df.rename(columns={'first': 'name'}, inplace=True)
```

```python
df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'])
df['age'] = pd.to_datetime(df['age'])
 
df['trans_date'] = df['trans_date_trans_time'].dt.date
df['trans_date'] = pd.to_datetime(df['trans_date'])  # Ensure it's in datetime format
```

```python
from dateutil.relativedelta import relativedelta
 
# Use apply with relativedelta for accurate age in years
df["age"] = df.apply(lambda row: relativedelta(row["trans_date"], row["age"]).years, axis=1)
 
```

```python
df['trans_month'] = pd.DatetimeIndex(df['trans_date']).month
df['trans_year'] = pd.DatetimeIndex(df['trans_date']).year
```

```python
df.drop(columns=['Unnamed: 0','trans_date_trans_time','unix_time','merch_lat','merch_long'], inplace=True)
```

```python
df.head()
```

## Data Visualization

### This plot shows the number of Fraud and Not Fraud limit.

```python
x = df['is_fraud'].unique()
y = df['is_fraud'].value_counts()
print(df.groupby('is_fraud')['is_fraud'].value_counts())

plt.figure(figsize=(8,6))
sns.set(style='darkgrid')

plt.bar(x,y,color=["lightblue", "coral"],data=df,label=['Fraud','Not Fraud'])
plt.yscale('log') 
plt.title('Fraud Class',fontsize=20)
plt.xlabel('Fraud',fontsize=15)
plt.ylabel('Count',fontsize=15)
plt.xticks([0,1], ['Not Fraud', 'Fraud'])
plt.legend(loc=1)
plt.show()
```

```python
# Separate features and labels
X = df.drop('is_fraud', axis=1)
y = df['is_fraud']
x
```

```python
# print(y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.utils import resample

train_data = pd.concat([X_train, y_train], axis=1)

not_fraud = train_data[train_data['is_fraud'] == 0]
fraud = train_data[train_data['is_fraud'] == 1]

sampled_df = resample(not_fraud, replace=False, n_samples=len(fraud), random_state=42)                                     
                                
sampled = pd.concat([sampled_df, fraud])

print(sampled['is_fraud'].value_counts())
```

```python
print(sampled['is_fraud'].value_counts())
sns.countplot(x='is_fraud', data=df)
sns.countplot(x='is_fraud', data=sampled)
plt.yscale('log')
```

```python
X_train_down = sampled.drop('is_fraud', axis=1)
y_train_down = sampled['is_fraud']
```

```python
X_train_down.shape, X_test.shape, y_train_down.shape, y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_down)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.ensemble  import RandomForestClassifier

model = RandomForestClassifier(random_state=42)
model.fit(X_train_scaled, y_train_down)
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, roc_auc_score

y_pred = model.predict(X_test_scaled)

print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nAccuracy Score:", accuracy_score(y_test, y_pred))
print("\nROC-AUC Score:", roc_auc_score(y_test, model.predict_proba(X_test_scaled)[:, 1]))
```
